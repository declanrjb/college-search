## IPEDs Tables

| Institutional Characteristics | Institution names, locations, websites, etc. Often joined to other tables. |
| ---- | ---- |
| 12-Month Enrollment | Headcount by race/ethnicity, gender, student level. |
| Fall Enrollment | Demographic info, age breakdowns, retention, student:faculty ratio, and more. |
| Cost | Application fees, tuition, food, housing, by institution by year. |
| Instructional Staff/Salaries | Grouped by gender and academic rank. |
| Admissions and test scores | Factors considered, applicants/admits/enrolled counts. |
| Graduation rates | % of students graduating within six years of entering (long lag time) |


## First Look Admissions Statistics

- Admission Rate: Admits / Applicants
    - How selective the school is
- Yield: Enrolled / Admits
    - How **in demand** the school is
- Low admission rate, low yield: very selective, but admitted students choose to go somewhere else (financial aid? Poached by other selective institutions?)
- High admission rate, high yield: Niche school. Fewer applicants, but those who do apply are committed.
- Low admission rate, high yield: Highly selective, high demand schools.
